```python
import numpy as np
import matplotlib.pyplot as plt
import quantecon as qe

# 设置图片清晰度，中文正常显示
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

# 定义生产函数：两种商品，面包和衣服
def production_possibility(x):
    # 资源总量固定，PPF曲线
    return np.sqrt(100 - x**2)

# 生成x轴数据：面包产量
x = np.linspace(0, 10, 200)
y = production_possibility(x)

# 绘图
plt.figure(figsize=(6,6))
plt.plot(x, y, color="#2266bb", linewidth=2.5, label="生产可能性边界 PPF")
plt.fill_between(x,y, alpha=0.15)

# 标注点
plt.scatter(6, np.sqrt(100-36), c="red", s=60, label="有效生产点(边界上)")
plt.scatter(4, 4, c="orange", s=60, label="无效率点(边界内部)")

plt.xlabel("面包产量")
plt.ylabel("衣服产量")
plt.title("生产可能性边界 PPF（微观经济学）")
plt.xlim(0,11)
plt.ylim(0,11)
plt.grid(alpha=0.3)
plt.legend()
plt.show()
```
